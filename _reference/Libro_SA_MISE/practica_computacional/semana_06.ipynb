## Capa 1. ¿Un promedio favorable resuelve la decisión?
Tres grupos ficticios valoran dos trazados. Los valores son unidades ilustrativas de una encuesta, no datos de comunidades reales ni una escala jurídica. Mira primero el promedio y después quién soporta el peor resultado.

```python
voces=pd.DataFrame({'Trazado A':[9,9,-8],'Trazado B':[4,4,3]},index=['Grupo 1','Grupo 2','Grupo 3'])
display(voces)
display(pd.DataFrame({'media_sin_pesos':voces.mean(),'minimo':voces.min()}))
```

El promedio no agota la distribución de cargas. Tampoco el mínimo decide por sí solo: faltan significado de la escala, representatividad, derechos y alternativas. La tabla muestra una pregunta que un único indicador oculta; no prescribe una regla para resolver conflictos.
