## Gobernanza desde la delimitación

La gobernanza no es un capítulo decorativo posterior al cálculo. Define quién decide, quién debe ser consultado, qué condiciones hacen admisible una alternativa y qué compromisos deben financiarse. La licencia ambiental, el proceso de consulta previa, el acceso a red, los contratos y la aceptación social son dimensiones relacionadas pero diferentes. Una no sustituye automáticamente a las demás.

Ostrom ayuda a estudiar arreglos con varios centros de autoridad. Para utilizar sus principios debe identificarse el recurso o problema de acción colectiva: no basta declarar que todo el SIN es un único bien común y asignarle calificaciones. Puede analizarse, por ejemplo, cómo se coordinan decisiones sobre un corredor, quién monitorea compromisos y qué mecanismo resuelve desacuerdos.

El análisis policéntrico no implica que muchos centros sean siempre mejores. Pueden aportar conocimiento y controles y también producir costos de coordinación, ambigüedad o exclusión. Hay que observar reglas concretas, capacidad de participación y evidencia del funcionamiento.

## Tres preguntas de justicia

La justicia distributiva examina quién recibe beneficios y soporta costos: pagos, calidad de servicio, ocupación del territorio, impactos ambientales y oportunidades. La procedimental pregunta quién accede a información, participa y puede cuestionar decisiones. La de reconocimiento examina si se respetan identidades, formas de organización y necesidades que una métrica agregada invisibiliza.

Un menor costo promedio de energía no demuestra mejora distributiva. Hacen falta beneficiarios y afectados identificados. Tampoco la existencia de una reunión demuestra participación efectiva. Conviene trabajar con indicadores observables y fuentes, conservando desacuerdos y datos pendientes.
