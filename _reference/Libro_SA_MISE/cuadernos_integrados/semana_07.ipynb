## Primera capa. Cuando cada especialista trae una respuesta correcta a otra pregunta

En una reunión, el analista financiero muestra una inversión rentable, el ingeniero señala una restricción de conexión y el equipo territorial presenta una condición pendiente. Nadie tiene necesariamente un cálculo equivocado. El problema puede ser que usaron fechas, fronteras o alternativas distintas. Integrar comienza por descubrir esas diferencias, no por promediar conclusiones.

Imagina que alguien dice «la alternativa cuesta menos». Pregunta respecto de qué, para quién, en qué fecha y qué dejó fuera. Si otra persona responde «es más robusta», pregunta bajo cuáles escenarios y qué métrica. Estas preguntas sencillas suelen mejorar un informe más que añadir otro gráfico.

## Segunda capa. Una cadena de afirmaciones que se pueda recorrer en ambos sentidos

Construiremos una cadena dato → transformación → resultado → mecanismo → decisión. Empezar por el dato permite comprobar unidades. Empezar por la recomendación permite preguntar qué evidencia la sostiene. Si una flecha falla, no basta con que los otros eslabones estén bien escritos.

Tomemos una conclusión docente: cierta política de almacenamiento reduce faltantes bajo un patrón sintético de aportes. La evidencia inmediata es una comparación con igual stock inicial y mismos aportes, cambiando solo la regla de operación. El mecanismo propuesto es conservar energía para después. Para recomendarla en una central real faltan restricciones y valoración del agua. La conclusión del experimento puede ser firme dentro del modelo sin convertirse en una orden de despacho.

La revisión entre pares debe buscar una explicación rival. Quizá el beneficio proviene de un activo adicional que no se cobró, de una condición inicial favorable o de haber cambiado dos parámetros. Reproducir el caso base y hacer una intervención controlada ayuda a discriminar esas posibilidades.

## Tercera capa. Diseñar el siguiente estudio también es una decisión

Una consultoría preliminar puede terminar recomendando qué estudiar primero. Para que eso sea útil, hay que precisar qué variable falta, qué alternativas separa y qué resultado haría cambiar la decisión. «Se necesitan más datos» es insuficiente. «Se necesita verificar esta restricción porque determina si esta alternativa es admisible» ya orienta una acción.

Ponderar métodos significa asignar a cada uno la pregunta que puede contestar. El flujo eléctrico examina factibilidad operativa; la caja, exposición del inversionista; el análisis institucional, condiciones y responsabilidades; el modelo dinámico, una hipótesis de interacción y demora. ## Una decisión, varias pruebas

Integrar no significa reunir gráficos. Una conclusión debe depender de resultados previos y mostrar sus conexiones. Ejemplo: menor producción o atraso de conexión → menor energía vendible → exposición contractual y caja → necesidad de revisar contrato o calendario. Si el grafo no informa el modelo financiero, declararlo; una coincidencia de títulos no constituye acoplamiento.

El caso del curso contiene dos escalas deliberadamente separadas: historia descriptiva del SIN y modelos ficticios para aprender mecanismos. Las cifras de una escala no validan automáticamente la otra. La integración profesional consiste en usar la primera para formular contexto/preguntas y la segunda para comprobar lógica, y después especificar los datos necesarios para el proyecto real.

## La afirmación como unidad de auditoría

Cada hallazgo debe tener identificador, fuente, fecha/universo, transformación, resultado y límite. El archivo de estado conserva estas propiedades por etapa. No usa textos de relleno para declarar completado un análisis faltante. Si una fuente cambia, se debe ejecutar la transformación y revisar las afirmaciones dependientes.

La suma de MW observados por tecnología puede diferir del total si faltan etiquetas. El correctivo es conservar «sin clasificar» y reconciliar, no omitir filas para producir una tabla limpia. Una tabla de escenarios también requiere una interpretación computada: el nombre de la ganadora debe salir del criterio y contemplar empates. Estas comprobaciones pequeñas evitan errores grandes en un informe convincente.

> **Alerta del consultor.** Reproducir el software y validar la inferencia son tareas distintas. Un pipeline que corre correctamente puede ejecutar una definición equivocada de margen de reserva. Por eso la revisión debe incluir una persona que explique el significado de cada indicador.

## Forma del informe

El memo ejecutivo tiene pregunta, destinatario, decisión propuesta o pendiente, principales razones, condiciones y siguiente acción. El cuerpo documenta fronteras, infraestructura, mercado, modelo, economía, gobernanza e incertidumbre. El anexo permite reproducir datos y cálculos; no es un depósito de figuras sin vínculo con las afirmaciones.

Como especificación operativa de esta edición se propone cuerpo de 5.000–7.000 palabras, con memo ejecutivo de una página y anexos reproducibles separados. Es una convención editorial de la revisión, no cambio de una aprobación institucional. Importa más cerrar la evidencia que alcanzar el máximo de palabras. El informe puede ser más breve si el encargo lo permite y el docente lo acuerda.

Una recomendación accionable identifica responsable con competencia, acción, secuencia, costo o estimación pendiente, plazo, condición habilitante y indicador de seguimiento. «Aumentar resiliencia» es un propósito; «solicitar el estudio de conexión actualizado para contrastar el escenario de atraso antes de comprometer el CAPEX» es una acción verificable.

## Revisión adversarial

El grupo revisor intenta cambiar la recomendación con una objeción concreta: unidad mal definida, falta de cobertura, supuesto monetario, alternativa omitida o competencia equivocada. El grupo autor responde con evidencia, modifica el análisis o limita la conclusión. Defender a toda costa el primer resultado no es un criterio de excelencia.

Se revisan por separado exactitud y relevancia. Una centralidad bien calculada puede no ser relevante para el contrato de energía que se está evaluando. Un VPN coherente puede depender de un precio capturado sin sustento. Un mapa institucional completo puede no explicar quién debe hacer la siguiente acción.

El umbral de una estrategia adaptativa necesita indicador observable, frecuencia, responsable y respuesta factible. No se inventa un porcentaje de embalse como gatillo nacional. Para el proyecto ficticio puede explorarse un umbral económico de precio capturado que vuelva negativo el VPN, dejando claro qué datos permitirían monitorearlo y qué cláusulas permitirían actuar.
