> **Alerta del consultor.** Una recomendación condicionada necesita condiciones que alguien pueda verificar. «Válida si no cambia el mundo» no sirve. «Revisar si la producción validada cae bajo X o el inicio comercial se retrasa Y, bajo este contrato y base monetaria» tiene contenido operativo.
